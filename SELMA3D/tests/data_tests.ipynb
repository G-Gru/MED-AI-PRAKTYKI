```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import io, color, data, exposure
import imagecodecs
```

```python
img = io.imread('test_pic.tif')
min_brightness = np.min(img)
max_brightness = np.max(img)
print(min_brightness, max_brightness)
fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(img, cmap='gray')
axes[0].set_title('Obraz')
axes[1].hist(img.ravel(), bins=256, range=(min_brightness, max_brightness))
axes[1].set_title('Histogram')
plt.yscale('log')
plt.show()
```

```python
p_low_val, p_high_val = np.percentile(img, (2, 98))
img_rescaled = exposure.rescale_intensity(img, in_range=(p_low_val, p_high_val))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(img_rescaled, cmap='gray')
axes[0].set_title('Obraz')
axes[1].hist(img_rescaled.ravel(), bins=256, range=(0, 67000))
plt.yscale('log')
axes[1].set_title('Histogram')
plt.show()
```
